```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.graph_objects as go
```

```python
df_oor = pd.read_csv('/Users/baselhussein/Projects/impossible_goals/agg/data/oor_clusters.csv')
df_bw = pd.read_csv('/Users/baselhussein/Projects/impossible_goals/agg/data/bw_clusters.csv')
df_cs = pd.read_csv('/Users/baselhussein/Projects/impossible_goals/agg/data/cs_clusters.csv')
df_qual = pd.read_csv('/Users/baselhussein/Projects/impossible_goals/agg/data/transition_counts-qual.csv')
```

```python
df_qual
```

```python
df_qual.drop(['count'], axis=1, inplace=True)
```

```python
df_qual.columns = ['cluster-category-oor', 'cluster-category-bw', 'cluster-category-cs', 'condition_1', 'condition_2']
```

```python
df_qual
```

```python
df_qual['cluster-category-oor'] = df_qual['cluster-category-oor'].str.replace('_oor', '', regex=False)
df_qual['cluster-category-bw'] = df_qual['cluster-category-bw'].str.replace('_bw', '', regex=False)
df_qual['cluster-category-cs'] = df_qual['cluster-category-cs'].str.replace('_cs', '', regex=False)
```

```python
df_bw_clean = df_bw[['ID', 'cluster-category-bw']]
df_bw_clean
```

```python
df_cs_clean = df_cs[['ID', 'cluster-category-cs']]
df_cs_clean
```

```python
df_oor_clean = df_oor[['ID', 'cluster-category-oor']]
df_oor_clean
```

```python
merged_df = pd.merge(df_oor_clean, df_bw_clean, on='ID', how='inner')
merged_df
```

```python
merged_df_final = pd.merge(merged_df, df_cs_clean, on='ID', how='inner')
merged_df_final
```

```python
df_qual
```

```python
# Clean the cluster-category columns in both DataFrames
for col in ['cluster-category-oor', 'cluster-category-bw', 'cluster-category-cs']:
    merged_df_final[col] = merged_df_final[col].str.strip().str.lower()
    df_qual[col] = df_qual[col].str.strip().str.lower()
```

```python
merged_df_final_qual = pd.merge(merged_df_final, df_qual, on=['cluster-category-oor', 'cluster-category-bw', 'cluster-category-cs'], how='inner')
```

```python
merged_df_final_qual
```

```python
merged_df_final_qual.rename(columns={
    'cluster-category-oor': 'out_of_reach',
    'cluster-category-bw': 'brick_wall',
    'cluster-category-cs': 'crab_storage'
}, inplace=True)
```

```python
merged_df_final_approach = pd.DataFrame(merged_df_final_qual['ID'])
merged_df_final_approach
```

```python
merged_df_final_approach['approach'] = (merged_df_final_qual['out_of_reach'] + " -> " + 
                              merged_df_final_qual['brick_wall'] + " -> " + 
                              merged_df_final_qual['crab_storage']
                              )
```

```python
# ids_evaluation = merged_df_final_approach['ID']
```
