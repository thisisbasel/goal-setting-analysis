```python
import pandas as pd 
import ast
```

```python
df = pd.read_csv("/Users/baselhussein/Projects/pps/data/Longform.csv")
```

```python
df
```

```python
df_levels_complete = df.dropna(subset=['Levels_complete'])
df_levels_complete
```

```python
def find_levels_completed(student): 
    longest_list = max(student['Levels_complete'], key=len)
    return pd.Series({
        #'count_levels_complete': len(longest_list),  
        'levels_complete': longest_list  
    })
```

```python
# Apply the function to each group of 'ID'
df_levels_complete_count = df_levels_complete.groupby('ID').apply(find_levels_completed).reset_index()
df_levels_complete_count
```

```python
# Function to convert strings to actual lists
def convert_to_list(row):
    if isinstance(row['levels_complete'], str):
        return ast.literal_eval(row['levels_complete'])
    return row['levels_complete']

# Apply the function to convert strings to lists
df_levels_complete_count['levels_complete'] = df_levels_complete_count.apply(convert_to_list, axis=1)
```

```python
# Function to count the number of items in a list
def count_items_in_list(row):
    return len(row['levels_complete'])

# Apply the function to count items in each list and create a new column 'list_length'
df_levels_complete_count['len_levels_complete'] = df_levels_complete_count.apply(count_items_in_list, axis=1)
```
